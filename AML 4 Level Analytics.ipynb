```python
import re
```

```python
def normalise_column_name(name: str) -> str:
    #1 Convert any input label to text and remove outer spaces.
    name = str(name).strip()
    #2 Replace groups of non-alphanumeric characters with one underscore.
    name = re.sub(r"[^0-9a-zA-Z]+", "_", name)
    #3 Collapse repeated underscores and remove underscores at both ends
    name = re.sub(r"_+", "_", name).strip("_")
    #4 Return lowercase snake_case for predictable downstream references.
    return name.lower()
```

```python
dataset['sender_entity'] = (dataset['from_bank'].astype(str).str.strip()
+ "::" + dataset['account'].astype(str).str.strip())
dataset['sender_entity']
```

```python
dataset['receiver_entity'] = (dataset['to_bank'].astype(str).str.strip() 
 + "::" + dataset['account_1'].astype(str).str.strip())
```

```python
dataset['transaction_id'] = (dataset.index.astype(str) + "-" + dataset['timestamp'].astype(str))
```

```python
dataset.head()
```

```python
# Sort transactions before calculating time-dependent behaviour.
dataset = dataset.sort_values(['sender_entity', 'timestamp']).copy()
dataset
```

```python
dataset['transaction_date'] = dataset['timestamp'].dt.date
```

```python
dataset['sender_daily_txn_count'] = dataset.groupby(['sender_entity', 'transaction_date'])['transaction_id'].transform("count")
```

```python
dataset
```

```python
dataset['sender_daily_amount'] = dataset.groupby(['sender_entity', 'transaction_date'])["amount_paid"].transform("sum")
```

```python
dataset['sender_daily_unique_receivers'] = dataset.groupby(['sender_entity', 'transaction_date'])['receiver_entity'].transform("nunique")
```

```python
dataset['previous_sender_timestamp'] = dataset.groupby('sender_entity')['timestamp'].shift(1)
dataset['previous_sender_timestamp']
```

```python
dataset['mins_since_previous_sender_transaction'] = (dataset['timestamp'] - dataset['previous_sender_timestamp']).dt.total_seconds() / 60.0
dataset['mins_since_previous_sender_transaction']
```

```python
dataset[dataset['mins_since_previous_sender_transaction'].between(0, 10, inclusive="both") == True]
```

```python
dataset["rapid_repeat_transfer"] = (
 dataset["mins_since_previous_sender_transaction"]
 .between(0, 10, inclusive="both")
 .fillna(False)
 .astype(int)
)
dataset.head()
```

```python
dataset['receiver_unique_senders'] = dataset.groupby('receiver_entity')['sender_entity'].transform("nunique")
```

```python
dataset['high_fan_in_receiver'] = (dataset['receiver_unique_senders'] >= 5).astype(int)
dataset.head()
```
